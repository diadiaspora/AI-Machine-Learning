# 5. You evaluated how good the model is

You calculated:

```python
mean_squared_error
r2_score
```

### Mean Squared Error (MSE)

Measures **how wrong predictions are**.

Smaller number = better.

Example:

```
MSE = 25  → good
MSE = 500 → bad
```

---

### R² Score

Shows **how much of the data variation your model explains**.

Range:

```
1.0  = perfect prediction
0.8  = very good
0.5  = moderate
0    = useless
```
