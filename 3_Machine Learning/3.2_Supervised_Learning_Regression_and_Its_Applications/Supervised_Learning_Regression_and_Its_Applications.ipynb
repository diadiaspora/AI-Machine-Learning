## Regression

![](https://labcontent.simplicdn.net/data-content/content-assets/Data_and_AI/ML/updated/Lesson_04/Linear.png)

__Independent variables(inputs)__

* a variable that is used to explain or predict another variable.
* These variables influence the prediction, but they are not the value being predicted.
* An independent variable is a predictor or input variable used to estimate the outcome in a regression model.

__Dependent variable (output)__

* the variable model is trying to predict or explain. Its value depends on the independent variables.
* A dependent variable is the outcome or target variable whose value is predicted using the independent variables.

### Application of Regression

1. Oil and Gas Industry: Various types of data are collected in the oil and gas industry from the surface and subsurface to understand production and sale processes. Linear and non-linear regression models forecast global oil production.

__Dependent Variable:__ Global oil production (what the model is trying to predict.)

__Inependent Variable:__ factors used to predict global oil production (Surface data, equipment performance, drilling activity, Subsurface data, reservoir pressure, rock properties, Historical production data, Oil prices, Number of active wells)

2. Marketing: In marketing, linear regression helps analyze the effectiveness of advertising campaigns, predict sales based on marketing spend, and segment customers based on demographic data.

__Dependent Variable:__ SALES 

__Inependent Variable:__ factors used to predict the outcome such as Marketing spend / advertising spend

3. Retail: Linear regression is utilized in retail for demand forecasting, inventory management, pricing optimization, and customer analytics.

__Dependent Variable:__ forecasting. Product demand (how many items will be sold)

__Inependent Variable:__ factors used to predict demand.(Product price, Promotions or discounts, Season or time of year, Historical sales data, Store location)

4. Healthcare: Linear regression is applied in healthcare for predicting patient outcomes, analyzing the relationship between medical variables and disease progression, and planning healthcare resources.

__Dependent Variable:__ Patient outcome (Recovery time, Disease progression level, Risk of complications, Length of hospital stay)

__Independent Variable:__ the medical factors used to make the prediction (Age, Blood pressure, Cholesterol level, Medical test results, Treatment type)

Real Estate: In the real estate industry, linear regression predicts property prices based on factors such as location, size, amenities, and economic indicators.

__Dependent Variable:__ Patient outcome (Recovery time, Disease progression level, Risk of complications, Length of hospital stay)

__Independent Variable:__ the medical factors used to make the prediction (Age, Blood pressure, Cholesterol level, Medical test results, Treatment type)
