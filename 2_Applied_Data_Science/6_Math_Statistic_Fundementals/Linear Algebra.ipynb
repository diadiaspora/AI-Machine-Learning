## `Python Code Snippet for Dot and Cross Product`

This function calculates the dot product of two vectors, not matrices.

A dot product is a way to multiply two vectors and get a single number as the result.

It measures how much two vectors point in the same direction.

```python
# Define a function to calculate dot product of two matrices
def dot_product(x,y):

    # Ensure that both the vectors have same length
    return sum(i*j for i,j in zip(x,y))
```

What are i and j?

They are just temporary variable names.

They represent:

i → one element from the first list (x)

j → the matching element from the second list (y)

```python
dot_product([3,2,6],[1,7,-2])
```

```python
# Define a function to calculate cross product of two vectors
def cross_product(u, v):
    q0 = u[1]*v[2] - u[2]*v[1]
    q1 = u[2]*v[0] - u[0]*v[2]
    q2 = u[0]*v[1] - u[1]*v[0]
    return [q0, q1, q2]

# Define vectors of same length
u = [1, 4, 5]
v = [2, -3, 6]

# Calculate the cross product
cross_product_result = cross_product(u, v)
cross_product_result
```

The cross product takes two 3D vectors and produces another 3D vector.

Unlike the dot product (which gives a single number), the cross product gives a new vector that is perpendicular to both original vectors.
